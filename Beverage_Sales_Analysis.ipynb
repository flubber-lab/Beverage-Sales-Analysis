```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.figure_factory as ff
import plotly.io as pio
```

```python
df = pd.read_csv("/content/synthetic_beverage_sales_data.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(axis=0, inplace=True)
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="%d-%m-%Y")
```

```python
df["Order_Date"].dtype
```

```python
df.describe()
```

```python
df.columns
```

```python
qty = df.groupby(["Product", "Category"])["Quantity"].sum().reset_index()
```

```python
pip install -U Kaleido
```

```python
fig = px.bar(qty, x ='Product', y='Quantity',color='Category',title='Sales Quantity by Product')
fig.show()
fig.write_image("Sales_quantity_by_product.png")
```

```python
fig1=px.pie(qty, values='Quantity', names='Category',hole=0.3 ,
       title='Sales Quantity by Category', color_discrete_sequence=px.colors.qualitative.Pastel)
fig1.show()
fig1.write_image("Sales_quantity_by_category.png")
```

```python
rgn = df.groupby(["Region","Category","Product"]).agg({'Quantity':'sum','Total_Price':'sum'}).reset_index()
rgn
```

```python
fig2 = px.bar(rgn, x='Region',y='Quantity',color='Category',barmode='stack'
,color_discrete_sequence=px.colors.qualitative.Pastel, title='Sales Quantity by Region')
fig2.show()
fig2.write_image("Sales_quantity_by_region.png")
```

```python
price = df.groupby(['Category','Product'])['Total_Price'].sum().reset_index()
fig3 = px.bar(price, x='Product', y='Total_Price', color='Category',
              title='Total Sales by Product', color_discrete_sequence=px.colors.qualitative.Pastel,
              labels={'Product':'Product','Total_Price':'Total Sales'})
fig3.show()
fig3.write_image("Total_sales_by_product.png")
```

```python
fig4=px.pie(price, names='Category', values='Total_Price', hole=0.3,
       title='Most Selling Category', color_discrete_sequence=px.colors.qualitative.Safe)
fig4.show()
fig4.write_image("Most_selling_category.png")
```
